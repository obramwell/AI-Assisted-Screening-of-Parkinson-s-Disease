# Multimodal Dataset Validation and Preparation
## Objective
This notebook validates the final participant-level datasets generated during the data integration stage and prepares them for multimodal machine learning experiments.

Rather than rebuilding the datasets, this notebook verifies that the corrected task-aware datasets are complete, internally consistent, and ready for hyperparameter tuning and model comparison.

The validated datasets will be used throughout Week 5 to develop and compare multimodal machine learning models.

## Datasets

The following participant-level datasets are loaded:

- Demographics + Questionnaire
- Wearable Features
- Wearable + Questionnaire
- Full Multimodal

These datasets were generated in the previous integration notebook using the corrected task-aware wearable features and the cleaned clinical data.

## Workflow

This notebook performs the following steps:

1. Load the finalized participant-level datasets.
2. Verify participant counts and duplicate identifiers.
3. Check missing values and feature completeness.
4. Confirm consistent participant-level records across datasets.
5. Validate train, validation, and test split assignments.
6. Produce summary tables describing each multimodal dataset.
7. Confirm that all datasets are ready for hyperparameter tuning and multimodal model development.

## 1. Libraries and configuration

Import the required libraries and locate the processed data folder containing the finalized participant-level datasets generated during Week 3.

```python
from pathlib import Path

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
```

## 2. Locate processed datasets

Locate the project root and define the paths to the finalized participant-level datasets used throughout Week 5.

```python
def find_project_root(start: Path) -> Path:
    """
    Locate the project root by searching for the data folder.
    """
    start = start.resolve()

    for candidate in [start, *start.parents]:
        if (candidate / "data" / "processed").exists():
            return candidate

    raise FileNotFoundError(
        "Could not locate the project's data/processed directory."
    )


project_root = find_project_root(Path.cwd())

processed_folder = project_root / "data" / "processed"

print("Project root:", project_root)
print("Processed folder:", processed_folder)
```

## 3. Dataset paths
Define the paths to the finalized task-aware participant-level datasets.

```python
demographics_questionnaire_file = (
    processed_folder /
    "demographics_questionnaire_task_aware.csv"
)

wearable_file = (
    processed_folder /
    "wearable_features_task_aware.csv"
)

wearable_questionnaire_file = (
    processed_folder /
    "wearable_questionnaire_task_aware.csv"
)

multimodal_file = (
    processed_folder /
    "multimodal_full_task_aware.csv"
)
```

## 4. Load finalized multimodal datasets
Load all participant-level datasets generated during the integration stage.

```python
demographics_questionnaire = pd.read_csv(
    demographics_questionnaire_file
)
wearable = pd.read_csv(
    wearable_file
)
wearable_questionnaire = pd.read_csv(
    wearable_questionnaire_file
)
multimodal = pd.read_csv(
    multimodal_file
)
print("All datasets loaded successfully.")
```

## 5. Verify dataset dimensions
Inspect the dimensions of each participant-level dataset before performing validation.

```python
dataset_summary = pd.DataFrame({
    "Dataset": [
        "Demographics + Questionnaire",
        "Wearable",
        "Wearable + Questionnaire",
        "Multimodal",
    ],
    "Participants": [
        demographics_questionnaire["patient_id"].nunique(),
        wearable["patient_id"].nunique(),
        wearable_questionnaire["patient_id"].nunique(),
        multimodal["patient_id"].nunique(),
    ],
    "Rows": [
        len(demographics_questionnaire),
        len(wearable),
        len(wearable_questionnaire),
        len(multimodal),
    ],
    "Columns": [
        demographics_questionnaire.shape[1],
        wearable.shape[1],
        wearable_questionnaire.shape[1],
        multimodal.shape[1],
    ],
})

display(dataset_summary)
```

## 6. Participant-level validation

```python
validation_summary = pd.DataFrame({
    "Dataset": [
        "Demographics + Questionnaire",
        "Wearable",
        "Wearable + Questionnaire",
        "Multimodal",
    ],
    "Unique Participants": [
        demographics_questionnaire["patient_id"].nunique(),
        wearable["patient_id"].nunique(),
        wearable_questionnaire["patient_id"].nunique(),
        multimodal["patient_id"].nunique(),
    ],
    "Duplicate IDs": [
        demographics_questionnaire["patient_id"].duplicated().sum(),
        wearable["patient_id"].duplicated().sum(),
        wearable_questionnaire["patient_id"].duplicated().sum(),
        multimodal["patient_id"].duplicated().sum(),
    ],
    "Missing Values": [
        demographics_questionnaire.isna().sum().sum(),
        wearable.isna().sum().sum(),
        wearable_questionnaire.isna().sum().sum(),
        multimodal.isna().sum().sum(),
    ]
})

display(validation_summary)
```

```python
validation_file = (
    processed_folder /
    "multimodal_dataset_validation_summary.csv"
)

validation_summary.to_csv(
    validation_file,
    index=False,
)

print("Validation summary saved.")
```

```python
assert all(
    validation_summary["Duplicate IDs"] == 0
)

print("PASS")
print("All participant-level datasets are ready for Week 5 modeling.")
```
